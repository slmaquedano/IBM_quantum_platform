# Certifying Steering on IBM Quantum Processors

We begin by providing a detailed description of the quantum circuit used to perform assemblage tomography. Based on protocols proposed in Refs. [1, 2], this circuit employs four qubits to generate Bell-diagonal (BD) states of the form:

$$
\rho_{CD} = \sum_{j,k}^1 p_{jk} \ket{\psi_{jk}}\bra{\psi_{jk}},
$$

where $\{p_{jk}\}_{j,k}$ is a set of probabilities summing to 1, and $\{\ket{\psi_{jk}}\}_{j,k}$ denotes the Bell states. The subcircuit responsible for generating these states is highlighted in purple in **Figure 1** (see below). 

The BD state is obtained through a purification process: the first two qubits ($q_A$ and $q_B$) act as an ancilla, and the state $\rho_{CD}$ is recovered by taking the partial trace over these qubits, i.e., $\rho_{CD} = \operatorname{tr}_{AB}(\rho)$, where $\rho$ is the global pure state of the system.

### State Preparation

The preparation of the initial state involves encoding the probabilities $\{p_{jk}\}_{j,k}$ into the first two qubits using rotation gates $R_y(\theta)\equiv \theta$. Specifically, the unitary operation $G$ prepares the state:

$$
G\ket{00}_{AB} = \sum_{j,k}^1\sqrt{p_{jk}}\ket{jk}_{AB}
$$

where the angles $\alpha$, $\beta$, and $\gamma$ are chosen to obtain the desired probability distribution. The demonstration for the explicit parameterization $(\alpha, \beta, \gamma) \rightarrow \{p_{jk}\}_{j,k}$ is presented in Ref. [2], resulting in:

$$
\begin{aligned}
\sqrt{p_{00}} &= \cos{\left(\frac{\alpha}{2}\right)}\cos{\left(\frac{\beta}{2}\right)}\cos{\left(\frac{\gamma}{2}\right)}+\sin{\left(\frac{\alpha}{2}\right)}\sin{\left(\frac{\beta}{2}\right)}\sin{\left(\frac{\gamma}{2}\right)}, \\
\sqrt{p_{01}} &= \cos{\left(\frac{\alpha}{2}\right)}\cos{\left(\frac{\beta}{2}\right)}\sin{\left(\frac{\gamma}{2}\right)}-\sin{\left(\frac{\alpha}{2}\right)}\sin{\left(\frac{\beta}{2}\right)}\cos{\left(\frac{\gamma}{2}\right)}, \\
\sqrt{p_{10}} &= \cos{\left(\frac{\alpha}{2}\right)}\sin{\left(\frac{\beta}{2}\right)}\cos{\left(\frac{\gamma}{2}\right)}-\sin{\left(\frac{\alpha}{2}\right)}\cos{\left(\frac{\beta}{2}\right)}\sin{\left(\frac{\gamma}{2}\right)}, \\
\sqrt{p_{11}} &= \cos{\left(\frac{\alpha}{2}\right)}\sin{\left(\frac{\beta}{2}\right)}\sin{\left(\frac{\gamma}{2}\right)}+\sin{\left(\frac{\alpha}{2}\right)}\cos{\left(\frac{\beta}{2}\right)}\cos{\left(\frac{\gamma}{2}\right)}.
\end{aligned}
$$

Subsequently, two CNOT gates map the state to $\sum_{j,k}^1\sqrt{p_{j,k}}\ket{jk}_{AB}\otimes\ket{jk}_{CD}$. Finally, a Hadamard gate followed by a CNOT operation transforms the states $\ket{jk}_{CD} \rightarrow \ket{\psi_{jk}}_{CD}$, resulting in the global state:

$$
\ket{\psi} = \sum_{j,k}^1\sqrt{p_{j,k}}\ket{jk}_{AB}\otimes\ket{\psi_{jk}}_{CD}.
$$

By taking the partial trace over qubits $q_A$ and $q_B$, we obtain the BD states.

### Werner States Configuration

For our analysis, we focus on a subset of BD states known as Werner states. These are characterized by the symmetry $p_{00}=p_{10}=p_{01}$ and $p_{11} = p_{00} + v$. These states are given by:

$$
\rho_{CD} = \frac{(1-v)}{4}\mathbb{I}\otimes\mathbb{I}+ v \ket{\psi_{11}}\bra{\psi_{11}},
$$

where $0\leq v\leq1$, and the singlet state $\ket{\psi_{11}}$ corresponds to $v = 1$. The configurations $(\alpha, \beta, \gamma)$ for specific visibilities used in this work are summarized in **Table 1**.

### Measurement Procedure

After preparing the Werner state, the subsystem $q_C$ (corresponding to the untrusted party) is measured in one of three bases $x \in \{\sigma_x, \sigma_y, \sigma_z\}$.

* **$\sigma_z$ basis:** Standard measurement in the computational basis.
* **$\sigma_x$ basis:** Requires applying a Hadamard gate before measurement.
* **$\sigma_y$ basis:** Requires applying a rotation $R_{z}(-\pi/2)$ followed by a Hadamard gate before measurement.

Once the basis change is applied, the measurement is performed, and the result is stored in the classical channel $c_A$. The subsystem $q_D$ (assumed to be the trusted party) is then subjected to **Quantum State Tomography**, a procedure that reconstructs the assemblage $\{\sigma_{a|x}\}_{a,x}$.
